최신 스토리 가져오기

```python
import requests
```

```python
BASE_URL = "https://hacker-news.firebaseio.com/v0/"

def fetch_data(endpoint):
    url = f"{BASE_URL}{endpoint}"
    response = requests.get(url)
    if response.status_code == 200:
        return response.json()
    else:
        raise Exception(f"Failed to fetch data. Status code: {response.status_code}")
```

```python
new_stories = fetch_data('newstories.json')
print(f'Number of new stories: {len(new_stories)}')
```

```python
new_stories
```

```python
def fetch_story_details(story_id):
    """Fetch detailed information for a specific story"""
    endpoint = f'item/{story_id}.json'
    story_data = fetch_data(endpoint)
    return story_data
```

```python
for story_id in new_stories[:5]:
    story_details = fetch_story_details(story_id)
    print(f'Title: {story_details.get('title')}')
    print(f'URL: {story_details.get('url')}')
    print(f'By: {story_details.get('by')}')
    print(f'Score: {story_details.get('score')}')
    print('-'*40)
```

사용자 정보 가져오기

```python
from datetime import datetime
```

```python
def fetch_user_details(username):
    """Fetch details of a specific user."""
    endpoint = f'user/{username}.json'
    user_data = fetch_data(endpoint)
    return user_data
```

```python
user_data = fetch_user_details('rduffyuk')
print(f'User: {user_data.get('id')}')
print(f'Created: {datetime.utcfromtimestamp(user_data.get('created')).strftime('%Y-%m-%d %H:%M:%S')}')
print(f'Karma: {user_data.get('karma')}')
print(f'about: {user_data.get('about')}')
```

최근 7일간의 인기스토리 가져오기

```python
from datetime import timedelta
```

```python
def filter_recent_stories(story_ids, days=7):
    """최근 특정 일수 이내의 스토리 필터링."""
    recent_stories = []
    current_time = datetime.now()
    time_limit = current_time - timedelta(days=days)

    for story_ids in story_ids:
        story_details = fetch_story_details(story_id)
        if story_details:
            story_time = datetime.utcfromtimestamp(story_details.get('time', 0))
            if story_time >= time_limit:
                recent_stories.append({
                    'id': story_id,
                    'title': story_details.get('title'),
                    'url': story_details.get('url'),
                    'time': story_time.strftime('%Y-%m-%d %H:%M:%S'),
                    'score': story_details.get('score'),
                    'by': story_details.get('by'),
                })
    return recent_stories

top_stories_ids = fetch_data('topstories.json')

if top_stories_ids:
    recent_stories = filter_recent_stories(top_stories_ids, days=7)

    print(f'Recent Top Stories (Last 7 Days): {len(recent_stories)} stories found')
    for index, story in enumerate(recent_stories, start=1):
        print(f'{index}')
        print(f'Title: {story['title']}')
        print(f'URL: {story['url']}')
        print(f'Time: {story['time']}')
        print(f'Score: {story['score']}')
        print(f'By: {story['by']}')
        print('-'*40)
```
